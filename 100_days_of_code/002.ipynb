# Day 2 - Data types, Numbers, Operations, Type conversion, f-strings

Task:

```bash
Welcome to the tip calculator
What was the total bill? $124.54
How much tip would you like to give? 10, 12, or 15? 12
How many people to split the bill? 5
Each person should pay $27.90
```

````python
print("Welcome to the tip calculator")
total_amount = 0
while total_amount == 0:
    try:
        total_amount = float(input("What was the total bill? $"))
    except ValueError:
        print("Please enter digits, not letters")

tip_percentage = 0
while tip_percentage == 0:
    try:
        tip_percentage = input("How much would you like to give? 10, 12, 15? ")
        tip_percentage = float('1.' + tip_percentage)
    except ValueError:
        print("Please enter digit amount")

split_between = int(input("How many people to split the bill?"))

split_pay = total_amount * tip_percentage / split_between
split_pay = round(split_pay, 2)
split_pay = f"{split_pay:.2f}"

print(f'''
 $ {total_amount:.2f}
+$  {total_amount * (tip_percentage-1):.2f}
_________  
 $ {total_amount * tip_percentage:.2f}

Amount to pay per person:
 $  {split_pay} 
''')
````
